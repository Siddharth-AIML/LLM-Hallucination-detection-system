```python
import json
import numpy as np
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
model = SentenceTransformer(
    "all-MiniLM-L6-v2"
)

print("Embedding model loaded")
```

```python
with open(
    "corpus.json",
    "r",
    encoding="utf-8"
) as f:

    corpus = json.load(f)

print("Corpus Size:", len(corpus))
```

```python
TOPICS = {

    "Fusion Reactor":
    "Nuclear fusion reactors, tokamak research, plasma energy, KSTAR reactor",

    "India AI Mission":
    "India artificial intelligence mission, AI summit, government AI initiatives",

    "Solar Storms":
    "Solar flares, geomagnetic storms, satellite communication disruption, space weather",

    "Quantum Computing":
    "Quantum computers, qubits, quantum supremacy, quantum processors",

    "CRISPR":
    "Gene editing, genome engineering, CRISPR technology, DNA modification"

}
```

```python
topic_embeddings = {}

for topic, description in TOPICS.items():

    embedding = model.encode(
        description
    )

    topic_embeddings[topic] = embedding

print("Topic embeddings created")
```

```python
topic_corpus = []

for topic in TOPICS:

    topic_corpus.append({

        "topic": topic,

        "description": TOPICS[topic],

        "evidence": []

    })
```

```python
SIMILARITY_THRESHOLD = 0.45

for sentence in corpus:

    sentence_embedding = model.encode(
        sentence
    )

    best_topic = None
    best_score = -1

    for topic in TOPICS:

        score = cosine_similarity(

            [sentence_embedding],

            [topic_embeddings[topic]]

        )[0][0]

        if score > best_score:

            best_score = score
            best_topic = topic

    if best_score >= SIMILARITY_THRESHOLD:

        for topic_obj in topic_corpus:

            if topic_obj["topic"] == best_topic:

                topic_obj["evidence"].append(
                    sentence
                )

                break
```

```python
for topic in topic_corpus:

    print("\n----------------")
    print(topic["topic"])

    print(
        len(topic["evidence"])
    )
```
