```python
import json
import re
from collections import defaultdict
```

```python
with open(
    "corpus.json",
    "r",
    encoding="utf-8"
) as f:

    corpus = json.load(f)

print("Corpus Size:", len(corpus))
```

```python
TOPIC_RULES = {

    "Fusion Reactor": [
        "fusion",
        "plasma",
        "tokamak",
        "reactor",
        "kstar"
    ],

    "India AI Mission": [
        "indiaai",
        "ai summit",
        "artificial intelligence",
        "india ai"
    ],

    "Solar Storms": [
        "solar flare",
        "solar storm",
        "satellite",
        "space weather"
    ],

    "Quantum Computing": [
        "quantum",
        "qubit",
        "quantum computing"
    ],

    "CRISPR": [
        "crispr",
        "gene editing",
        "genome"
    ]
}
```

```python
topic_corpus = []

for topic_name, keywords in TOPIC_RULES.items():

    evidence = []

    for sentence in corpus:

        sentence_lower = sentence.lower()

        for keyword in keywords:

            if keyword in sentence_lower:

                evidence.append(sentence)

                break

    topic_corpus.append({

        "topic": topic_name,

        "keywords": keywords,

        "evidence": evidence

    })
```

```python
for topic in topic_corpus:

    print("\n================")
    print("TOPIC:", topic["topic"])

    print(
        "Evidence Count:",
        len(topic["evidence"])
    )
```

```python
with open(
    "topic_corpus.json",
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        topic_corpus,
        f,
        indent=4,
        ensure_ascii=False
    )

print("Saved topic_corpus.json")
```
