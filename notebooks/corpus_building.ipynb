```python
def fetch_wikipedia_data(topics):

    corpus_raw = []

    for topic in topics:
        try:
            print(f"Fetching: {topic}")
            page = wikipedia.page(topic)
            corpus_raw.append(page.content)
        except Exception as e:
            print(f"❌ Error fetching {topic}: {e}")

    return corpus_raw


raw_texts = fetch_wikipedia_data(topics)
```

```python
def clean_text(text):

    # remove references like [1], [2]
    text = re.sub(r'\[\d+\]', '', text)

    # remove extra spaces
    text = re.sub(r'\s+', ' ', text)

    return text.strip()


cleaned_texts = [clean_text(t) for t in raw_texts]
```

```python
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
def split_into_sentences(texts):

    sentences = []

    for text in texts:
        doc = nlp(text)
        for sent in doc.sents:
            sentences.append(sent.text.strip())

    return sentences


corpus = split_into_sentences(cleaned_texts)

print("Total sentences:", len(corpus))
```

```python
def filter_sentences(sentences):

    filtered = []

    for s in sentences:

        # remove very short sentences
        if len(s.split()) < 6:
            continue

        # remove noisy lines
        if any(x in s.lower() for x in [
            "may refer to", "list of", "see also"
        ]):
            continue

        filtered.append(s.strip())

    return filtered


corpus = filter_sentences(corpus)

print("Filtered corpus size:", len(corpus))
```

```python
import json

with open("corpus.json", "w", encoding="utf-8") as f:
    json.dump(corpus, f, indent=2)

print("Corpus saved as corpus.json")
```
